Task 1.a

```python
import numpy as np

def d_e(v_0, v_mean):
    return (v_0 - v_mean).T@(v_0 - v_mean)


def d_m(v_0, v_mean, S):
    return (v_0 - v_mean).T@np.linalg.inv(S)@(v_0 - v_mean)
```

```python
v_mean = np.array([[0], [1], [1]])
v_0 = np.array([[2], [5], [3]])
S = np.array([[5, 1, 2],
              [1, 6, 1],
              [2, 1, 7]])
```

```python
print(f"d_e = {d_e(v_0, v_mean)}")
```

```python
print(f"d_m = {d_m(v_0, v_mean, S)}")
```

```python
import pandas as pd

df = pd.read_csv('dataFig311.csv')
df
```

```python
import statsmodels.api as sm

X = df['educ']
y = df['lnwage']
X = sm.add_constant(X)
model = sm.OLS(y, X).fit()
print(model.summary())
```

```python
df_agg = df.groupby(by='educ', as_index=False).agg(
    mean =('lnwage', 'mean'),
    cnt =('lnwage', 'size')
)

df_agg
```

```python
X = df_agg['educ']
y = df_agg['mean']
weights = df_agg['cnt']
X = sm.add_constant(X)
model = sm.WLS(y, X, weights=weights).fit()
print(model.summary())
```
